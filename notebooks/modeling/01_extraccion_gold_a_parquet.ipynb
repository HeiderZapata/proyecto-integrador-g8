Este notebook sirve para descargar el dataset qeu ya esta en arquitectura Gold como un parquet comprimido para realizar 

```python
gold = spark.read.format("delta").load("/Volumes/workspace/default/e_commerce/gold/features_session")
(gold.write.mode("overwrite")
     .option("compression", "zstd")
     .parquet("/Volumes/workspace/default/e_commerce/gold_snapshot"))
```

Verificación de que el parquet quedo bien y se puede exportar

```python
snap = spark.read.parquet("/Volumes/workspace/default/e_commerce/gold_snapshot")
print(f"{snap.count():,} filas  ×  {len(snap.columns)} columnas")   # esperado ~22.99M × 22
```
